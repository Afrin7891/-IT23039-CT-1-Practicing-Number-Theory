```python
def extended_gcd(a, b):
    assert a >= b and b >= 0 and a + b > 0

    d, x, y = 0, 0, 0

    if b == 0:
        d = a
        x = 1
        y = 0
    else:
        d, x1, y1 = extended_gcd(b, a % b)

        x = y1
        y = x1 - (a // b) * y1

    assert a % d == 0 and b % d == 0
    assert d == a * x + b * y

    return (d, x, y)
```

```python
print(extended_gcd(8, 4))
print(extended_gcd(27, 15))
```

```python
def lcm(a, b):
    d, x, y = extended_gcd(max(a, b), min(a, b))

    return (a // d) * b
```

```python
assert lcm(9, 3) == 9
assert lcm(17, 37) == 629

print(lcm(9, 3))
print(lcm(17, 37))
```

```python
def diophantine(a, b, c):

    if a >= b:
        gcd, x1, y1 = extended_gcd(a, b)
        x, y = x1, y1
    else:
        gcd, x1, y1 = extended_gcd(b, a)
        x, y = y1, x1

    x *= c // gcd
    y *= c // gcd

    assert c % gcd == 0
    assert a * x + b * y == c

    return (x, y)
```

```python
print(diophantine(34, 51, 170))
print(diophantine(91, 117, 65))
```

```python
def modInverse(a, n):
    d, x, y = extended_gcd(max(a, n), min(a, n))

    if a >= n:
        inverse = x
    else:
        inverse = y

    return inverse % n


def divide(b, a, n):
    assert n > 1 and a > 0

    aInv = modInverse(a, n)

    assert (a * aInv) % n == 1

    return (aInv * b) % n
```

```python
print(divide(9, 5, 3))
print(divide(223, 67, 54))
```

```python
def FastModularExponentiation(b, e, m):

    result = 1
    b = b % m

    while e > 0:

        if e % 2 == 1:
            result = (result * b) % m

        b = (b * b) % m
        e = e // 2

    return result
```

```python
print(FastModularExponentiation(4, 5, 3))
```

```python
def ChineseRemainderTheorem(n1, r1, n2, r2):

    d, x, y = extended_gcd(n1, n2)

    assert d == 1

    result = (r1 * n2 * y + r2 * n1 * x)

    return result % (n1 * n2)
```

```python
print(ChineseRemainderTheorem(13, 7, 6, 5))
print(ChineseRemainderTheorem(28, 20, 19, 3))
```

```python
def repunit(n):
    number = 0

    for i in range(n):
        number = number * 10 + 1

    return number


print(repunit(1))
print(repunit(2))
print(repunit(3))
print(repunit(4))
```

```python
def is_prime(n):

    if n < 2:
        return False

    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False

    return True
```

```python
def find_Ap(p, limit):

    if not is_prime(p):
        return "p must be prime"

    result = []

    for n in range(1, limit + 1):

        if repunit(n) % p == 0:
            result.append(n)

    return result


print("A_3 =", find_Ap(3, 30))
print("A_7 =", find_Ap(7, 30))
print("A_11 =", find_Ap(11, 30))
```

```python
def is_mersenne_prime(n):

    if not is_prime(n):
        return False

    mersenne = 2**n - 1

    return is_prime(mersenne)


print(is_mersenne_prime(2))
print(is_mersenne_prime(3))
print(is_mersenne_prime(4))
print(is_mersenne_prime(5))
```

```python
def even_perfect_numbers(limit):

    result = []

    for n in range(2, limit + 1):

        if is_mersenne_prime(n):

            k = (2**(n - 1)) * (2**n - 1)

            result.append(k)

    return result


print(even_perfect_numbers(10))
```

```python
def is_perfect(k):

    if k <= 1:
        return False

    divisor_sum = 0

    for i in range(1, k):
        if k % i == 0:
            divisor_sum += i

    return divisor_sum == k


print(is_perfect(6))
print(is_perfect(28))
print(is_perfect(12))
print(is_perfect(496))
```

```python
def is_composite(n):

    if n <= 3:
        return False

    return not is_prime(n)
```

```python
def is_fermat_pseudoprime(n, a):

    if n <= 1:
        return False

    if not is_composite(n):
        return False

    if gcd(a, n) != 1:
        return False

    return pow(a, n - 1, n) == 1
```

```python
def gcd(a, b):

    while b != 0:
        a, b = b, a % b

    return a
```

```python
def find_pseudoprimes(a, limit):

    result = []

    for n in range(2, limit + 1):

        if is_fermat_pseudoprime(n, a):
            result.append(n)

    return result


print("Base 2:")
print(find_pseudoprimes(2, 2000))

print("Base 3:")
print(find_pseudoprimes(3, 2000))
```
